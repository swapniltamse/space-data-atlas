# Find CRISM mineral cubes near Jezero

Dataset: [MRO CRISM Mineral Map Products (MTRDR)](https://pds-geosciences.wustl.edu/missions/mro/crism.htm). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#crism-mtrdr).

Independent project. Not affiliated with, produced by, or endorsed by NASA or NASA Space Apps.

```python
import requests

# Orbital Data Explorer search: CRISM map-projected targeted cubes (MTRDR)
r = requests.get("https://oderest.rsl.wustl.edu/live2/", params={
    "query": "product", "results": "fp", "output": "JSON", "limit": 5,
    "target": "mars", "ihid": "MRO", "iid": "CRISM", "pt": "MTRDR",
    "westlon": 77, "eastlon": 78, "minlat": 18, "maxlat": 19,  # Jezero Crater area
}, timeout=60)
r.raise_for_status()
for p in r.json()["ODEResults"]["Products"]["Product"]:
    files = p["Product_files"]["Product_file"]
    img = [f["URL"] for f in files if f["URL"].endswith(".img")]
    print(p["pdsid"], "|", len(img), "image cubes")
```
