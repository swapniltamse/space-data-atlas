# Search ECOSTRESS evapotranspiration tiles

Dataset: [ECOSTRESS Evapotranspiration (ECO_L3T_JET) V003](https://cmr.earthdata.nasa.gov/search/concepts/C4057232489-LPCLOUD.html). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#ecostress-et).

```python
import earthaccess

# Search needs no login. Bounding box: (west, south, east, north)
results = earthaccess.search_data(
    short_name="ECO_L3T_JET",
    bounding_box=(-121.0, 36.5, -119.5, 37.5),  # Central Valley, California
    temporal=("2026-06-01", "2026-10-15"),
    count=5,
)
print(len(results), "granules")
for g in results:
    print(g["umm"]["GranuleUR"])

# Next step (downloads need a free Earthdata Login):
# earthaccess.login()
# files = earthaccess.download(results[:1], "./ecostress")
# import rasterio; img = rasterio.open(files[0]).read(1)
```
