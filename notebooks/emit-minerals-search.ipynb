# Search EMIT surface mineral maps

Dataset: [EMIT L2B Surface Mineral Identification (EMITL2BMIN)](https://cmr.earthdata.nasa.gov/search/concepts/C2408034484-LPCLOUD.html). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#emit-mineralogy).

```python
import earthaccess

# Search needs no login. Bounding box: (west, south, east, north)
results = earthaccess.search_data(
    short_name="EMITL2BMIN",
    bounding_box=(-116.5, 35.0, -115.5, 36.0),  # Mojave Desert
    temporal=("2024-01-01", "2026-10-15"),
    count=5,
)
print(len(results), "granules")
for g in results:
    print(g["umm"]["GranuleUR"])

# Next step (downloads need a free Earthdata Login):
# earthaccess.login()
# files = earthaccess.download(results[:1], "./emit")
# import xarray as xr; ds = xr.open_dataset(files[0])
```
