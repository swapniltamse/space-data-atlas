# Pull fire detections for a bounding box

Dataset: [FIRMS Active Fire API](https://firms.modaps.eosdis.nasa.gov/api/area/). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#firms-api).

Independent project. Not affiliated with, produced by, or endorsed by NASA or NASA Space Apps.

```python
import os, io
import requests
import pandas as pd

# Get a free MAP_KEY at https://firms.modaps.eosdis.nasa.gov/api/map_key/
key = os.environ["FIRMS_MAP_KEY"]
source = "VIIRS_SNPP_SP"          # or MODIS_SP for the MODIS archive
bbox = "-124.5,32.5,-114.0,42.0"  # west,south,east,north
url = f"https://firms.modaps.eosdis.nasa.gov/api/area/csv/{key}/{source}/{bbox}/5/2025-08-01"

r = requests.get(url, timeout=60)
r.raise_for_status()
df = pd.read_csv(io.StringIO(r.text))
print(len(df), "detections")
print(df[["latitude", "longitude", "acq_date", "confidence"]].head())
```
