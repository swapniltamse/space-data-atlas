# Find HiRISE images of Jezero Crater

Dataset: [HiRISE Images (MRO)](https://www.uahirise.org/). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#hirise).

Independent project. Not affiliated with, produced by, or endorsed by NASA or NASA Space Apps.

```python
import requests

# Orbital Data Explorer search: HiRISE map-projected images in a lat/lon box
r = requests.get("https://oderest.rsl.wustl.edu/live2/", params={
    "query": "product", "results": "fp", "output": "JSON", "limit": 5,
    "target": "mars", "ihid": "MRO", "iid": "HIRISE", "pt": "RDRV11",
    "westlon": 77.3, "eastlon": 77.6, "minlat": 18.3, "maxlat": 18.6,  # Jezero Crater
}, timeout=60)
r.raise_for_status()
for p in r.json()["ODEResults"]["Products"]["Product"]:
    files = p["Product_files"]["Product_file"]
    thumb = next((f["URL"] for f in files if f["FileName"].endswith(".THUMB.JPG")), None)
    print(p["pdsid"], "|", thumb)
```
