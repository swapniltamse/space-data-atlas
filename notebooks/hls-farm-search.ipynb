# Find low-cloud HLS scenes over a farm

Dataset: [Harmonized Landsat Sentinel-2 (HLS) Landsat 30 m (HLSL30) v2.0](https://www.earthdata.nasa.gov/data/catalog/lpcloud-hlsl30-2.0). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#hls-landsat-l30).

```python
import earthaccess

# 30 m HLS Landsat scenes over a single farm area, low cloud only
results = earthaccess.search_data(
    short_name="HLSL30",
    bounding_box=(-93.7, 41.9, -93.5, 42.1),  # near Ames, Iowa
    temporal=("2026-05-01", "2026-09-30"),
    cloud_cover=(0, 20),
    count=5,
)
print(len(results), "scenes")
for g in results:
    print(g["umm"]["GranuleUR"])

# Each scene has one Cloud Optimized GeoTIFF per band.
# earthaccess.login()
# earthaccess.download(results[:1], "./hls")
```
