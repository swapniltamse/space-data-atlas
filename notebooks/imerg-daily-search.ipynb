# Search daily GPM IMERG rainfall

Dataset: [GPM IMERG Final Precipitation L3 Daily 0.1 deg (GPM_3IMERGDF) V07](https://disc.gsfc.nasa.gov/datasets/GPM_3IMERGDF_07/summary). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#gpm-imerg-daily).

```python
import earthaccess

# Search needs no login. This product is global, so no bounding box.
results = earthaccess.search_data(
    short_name="GPM_3IMERGDF",
    temporal=("2025-09-01", "2025-09-10"),
    count=5,
)
print(len(results), "granules")
for g in results:
    print(g["umm"]["GranuleUR"])

# Next step (downloads need a free Earthdata Login):
# earthaccess.login()
# files = earthaccess.download(results[:1], "./imerg")
# import xarray as xr; ds = xr.open_dataset(files[0])
# The Final Run lags by months. For recent days, use short_name="GPM_3IMERGDL" (Late Run).
```
