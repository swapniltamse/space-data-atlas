# List LOLA lunar elevation maps

Dataset: [LRO LOLA Gridded Topography via PDS Orbital Data Explorer REST API](https://oderest.rsl.wustl.edu/). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#lola-ode-rest).

Independent project. Not affiliated with, produced by, or endorsed by NASA or NASA Space Apps.

```python
import requests

# Ask the Orbital Data Explorer for LOLA gridded elevation products
r = requests.get("https://oderest.rsl.wustl.edu/live2/", params={
    "query": "product", "results": "fp", "output": "JSON", "limit": 5,
    "target": "moon", "ihid": "LRO", "iid": "LOLA", "pt": "GDRDEM",
}, timeout=60)
r.raise_for_status()
for p in r.json()["ODEResults"]["Products"]["Product"]:
    files = p["Product_files"]["Product_file"]
    jpg = next((f["URL"] for f in files if f["URL"].endswith(".jpg")), None)
    print(p["pdsid"], "| preview:", jpg)
```
