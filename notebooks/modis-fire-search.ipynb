# Search MODIS Terra fire detections

Dataset: [MODIS/Terra Thermal Anomalies/Fire 5-Min L2 Swath 1km (MOD14) V061](https://www.earthdata.nasa.gov/data/catalog/lpcloud-mod14-061). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#modis-terra-fire-mod14).

```python
import earthaccess

# Search needs no login. Bounding box: (west, south, east, north)
results = earthaccess.search_data(
    short_name="MOD14",
    bounding_box=(-124.4, 32.5, -114.1, 42.0),  # California
    temporal=("2026-09-01", "2026-10-15"),
    count=5,
)
print(len(results), "granules")
for g in results:
    print(g["umm"]["GranuleUR"])

# Next step (downloads need a free Earthdata Login):
# earthaccess.login()
# files = earthaccess.download(results[:1], "./mod14")
```
