# Search 16-day MODIS vegetation index

Dataset: [MODIS/Terra Vegetation Indices 16-Day L3 Global 250m (MOD13Q1) V061](https://www.earthdata.nasa.gov/data/catalog/lpcloud-mod13q1-061). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#modis-ndvi-mod13q1).

```python
import earthaccess

# Search needs no login. Bounding box: (west, south, east, north)
results = earthaccess.search_data(
    short_name="MOD13Q1",
    bounding_box=(-96.6, 40.4, -90.1, 43.5),  # Iowa farmland
    temporal=("2026-06-01", "2026-10-15"),
    count=5,
)
print(len(results), "granules")
for g in results:
    print(g["umm"]["GranuleUR"])

# Next step (downloads need a free Earthdata Login):
# earthaccess.login()
# files = earthaccess.download(results[:1], "./ndvi")
```
