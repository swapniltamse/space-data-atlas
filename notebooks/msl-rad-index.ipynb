# Read the Curiosity RAD file index

Dataset: [Curiosity RAD Surface Radiation Data](https://pds-ppi.igpp.ucla.edu/data/MSL-M-RAD-3-RDR-V1.0/). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#msl-rad).

Independent project. Not affiliated with, produced by, or endorsed by NASA or NASA Space Apps.

```python
import requests

# The PDS index lists every RAD radiation file, one per row
base = "https://pds-ppi.igpp.ucla.edu/data/MSL-M-RAD-3-RDR-V1.0/"
rows = requests.get(base + "INDEX/INDEX.TAB", timeout=60).text.splitlines()
print(len(rows), "files in the index")
for row in rows[:5]:
    print(row[:120])
# Each row has the file path; prefix it with the base URL to download it
```
