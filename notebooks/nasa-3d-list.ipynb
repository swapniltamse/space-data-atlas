# List NASA 3D models

Dataset: [NASA 3D Resources](https://github.com/nasa/NASA-3D-Resources). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#nasa-3d-resources).

Independent project. Not affiliated with, produced by, or endorsed by NASA or NASA Space Apps.

```python
import requests

# GitHub's API lists the model folders. Unauthenticated calls are limited
# to 60 an hour per network, so cache the list rather than calling in a loop.
url = "https://api.github.com/repos/nasa/NASA-3D-Resources/contents/3D%20Models"
models = requests.get(url, timeout=30).json()
print(len(models), "models")
for m in models[:10]:
    print(m["name"], "|", m["html_url"])
```
