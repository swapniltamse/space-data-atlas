# Search NISAR interferograms

Dataset: [NISAR L2 Geocoded Unwrapped Interferogram (GUNW), Provisional](https://www.earthdata.nasa.gov/data/catalog/asf-nisar-l2-gunw-provisional-v1-1). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#nisar-l2-gunw).

```python
import earthaccess

# Search needs no login. Bounding box: (west, south, east, north)
results = earthaccess.search_data(
    short_name="NISAR_L2_GUNW_PROVISIONAL_V1",
    bounding_box=(-122.6, 37.0, -121.5, 38.2),  # San Francisco Bay
    temporal=("2025-10-01", "2026-10-15"),
    count=5,
)
print(len(results), "granules")
for g in results:
    print(g["umm"]["GranuleUR"])

# Next step (downloads need a free Earthdata Login):
# earthaccess.login()
# files = earthaccess.download(results[:1], "./nisar")
```
