# Search fire safety reports

Dataset: [NASA Technical Reports Server (NTRS) API](https://ntrs.nasa.gov/api/openapi/). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#ntrs-api).

Independent project. Not affiliated with, produced by, or endorsed by NASA or NASA Space Apps.

```python
import requests

# Search NASA technical reports on microgravity combustion
params = {"q": "microgravity flame spread spacecraft fire safety", "page.size": 5}
r = requests.get("https://ntrs.nasa.gov/api/citations/search", params=params, timeout=60)
r.raise_for_status()
data = r.json()
print("total matches:", data["stats"]["total"])
for rec in data["results"]:
    print(rec["id"], "|", rec["title"][:90])
    # Record page: https://ntrs.nasa.gov/citations/<id>
```
