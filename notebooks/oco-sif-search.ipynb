# Search OCO-2 plant fluorescence (SIF)

Dataset: [OCO-2 Lite Solar-Induced Fluorescence (OCO2_L2_Lite_SIF)](https://cmr.earthdata.nasa.gov/search/concepts/C2912084771-GES_DISC.html). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#oco-sif).

```python
import earthaccess

# Search needs no login. This product is global, so no bounding box.
results = earthaccess.search_data(
    short_name="OCO2_L2_Lite_SIF",
    temporal=("2026-07-01", "2026-07-27"),
    count=5,
)
print(len(results), "granules")
for g in results:
    print(g["umm"]["GranuleUR"])

# Next step (downloads need a free Earthdata Login):
# earthaccess.login()
# files = earthaccess.download(results[:1], "./oco_sif")
# import xarray as xr; ds = xr.open_dataset(files[0])
```
