# Search OPERA land disturbance alerts

Dataset: [OPERA Land Surface Disturbance Alert from HLS (DIST-ALERT)](https://www.earthdata.nasa.gov/data/catalog/lpcloud-opera-l3-dist-alert-hls-v1-1). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#opera-dist-alert-hls).

```python
import earthaccess

# Search needs no login. Bounding box: (west, south, east, north)
results = earthaccess.search_data(
    short_name="OPERA_L3_DIST-ALERT-HLS_V1",
    bounding_box=(-63.5, -10.5, -62.5, -9.5),  # Rondonia, Brazil
    temporal=("2026-08-01", "2026-10-15"),
    count=5,
)
print(len(results), "granules")
for g in results:
    print(g["umm"]["GranuleUR"])

# Next step (downloads need a free Earthdata Login):
# earthaccess.login()
# files = earthaccess.download(results[:1], "./opera_dist")
# import rasterio; img = rasterio.open(files[0]).read(1)
```
