# Search OPERA surface water maps

Dataset: [OPERA Dynamic Surface Water Extent from HLS (DSWx-HLS)](https://podaac.jpl.nasa.gov/dataset/OPERA_L3_DSWX-HLS_V1). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#opera-dswx-hls).

```python
import earthaccess

# Search needs no login. Bounding box: (west, south, east, north)
results = earthaccess.search_data(
    short_name="OPERA_L3_DSWX-HLS_V1",
    bounding_box=(-81.2, 26.6, -80.6, 27.2),  # Lake Okeechobee, Florida
    temporal=("2026-09-01", "2026-10-15"),
    count=5,
)
print(len(results), "granules")
for g in results:
    print(g["umm"]["GranuleUR"])

# Next step (downloads need a free Earthdata Login):
# earthaccess.login()
# files = earthaccess.download(results[:1], "./opera_dswx")
# import rasterio; img = rasterio.open(files[0]).read(1)
```
