# Search spaceflight bone-loss studies

Dataset: [NASA Open Science Data Repository (OSDR) Search API](https://osdr-tutorials.readthedocs.io/en/latest/pages/guides/biological_data_api.html). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#nasa-osdr).

Independent project. Not affiliated with, produced by, or endorsed by NASA or NASA Space Apps.

```python
import requests

# Search public spaceflight biology studies (GeneLab + ALSDA)
params = {"term": "bone loss", "type": "cgene,alsda", "from": 0, "size": 5}
r = requests.get("https://osdr.nasa.gov/osdr/data/search", params=params, timeout=60)
r.raise_for_status()
hits = r.json()["hits"]
print("matching studies:", hits["total"])
for h in hits["hits"]:
    s = h["_source"]
    print(s.get("Accession"), "|", s.get("Study Title", "")[:80])
```
