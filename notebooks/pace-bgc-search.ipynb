# Search PACE ocean chlorophyll maps

Dataset: [PACE OCI Ocean Biogeochemistry (L3M BGC)](https://cmr.earthdata.nasa.gov/search/concepts/C4184125847-OB_CLOUD.html). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#pace-oci-bgc).

```python
import earthaccess

# Search needs no login. This product is global, so no bounding box.
results = earthaccess.search_data(
    short_name="PACE_OCI_L3M_BGC",
    temporal=("2026-08-01", "2026-08-31"),
    count=5,
)
print(len(results), "granules")
for g in results:
    print(g["umm"]["GranuleUR"])

# Next step (downloads need a free Earthdata Login):
# earthaccess.login()
# files = earthaccess.download(results[:1], "./pace")
# import xarray as xr; ds = xr.open_dataset(files[0])
```
