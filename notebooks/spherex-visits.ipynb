# List SPHEREx visits to one sky position

Dataset: [SPHEREx Quick Release Images at IRSA (TAP / SIA)](https://irsa.ipac.caltech.edu/Missions/spherex.html). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#irsa-spherex).

Independent project. Not affiliated with, produced by, or endorsed by NASA or NASA Space Apps.

```python
import io
import requests
import pandas as pd

# Find every SPHEREx image (detector D1) that covers one sky position.
# Repeat visits of the same spot are what reveal moving objects.
ra, dec = 150.1, 2.2  # COSMOS field; change to your target
query = f"""
SELECT obs_id, obs_collection, t_min, access_url
FROM spherex.obscore
WHERE CONTAINS(POINT('ICRS', {ra}, {dec}), s_region) = 1
  AND energy_bandpassname = 'SPHEREx-D1'
ORDER BY t_min
"""
r = requests.get("https://irsa.ipac.caltech.edu/TAP/sync",
                 params={"QUERY": query, "FORMAT": "csv"}, timeout=300)
df = pd.read_csv(io.StringIO(r.text))
print(len(df), "visits")  # t_min is a Modified Julian Date
print(df[["obs_id", "obs_collection", "t_min"]].head())
```
