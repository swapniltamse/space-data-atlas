# Load a SPICE kernel and convert a date

Dataset: [NAIF SPICE Toolkit and Kernels](https://naif.jpl.nasa.gov/naif/toolkit.html). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#naif-spice).

```python
import requests
import spiceypy  # pip install spiceypy

# The leap-seconds kernel is the first thing almost every SPICE program loads
url = "https://naif.jpl.nasa.gov/pub/naif/generic_kernels/lsk/naif0012.tls"
open("naif0012.tls", "wb").write(requests.get(url, timeout=60).content)
spiceypy.furnsh("naif0012.tls")

et = spiceypy.str2et("2026-11-15T12:00:00")
print("seconds past J2000 (TDB):", et)
# Next: load planet and spacecraft kernels (spk/, pck/) and call spiceypy.spkpos
```
