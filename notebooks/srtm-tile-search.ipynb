# Find SRTM elevation tiles for an area

Dataset: [NASA SRTM Global 1 arc second (SRTMGL1) V003](https://www.earthdata.nasa.gov/data/catalog/lpcloud-srtmgl1-003). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#srtm-elevation).

```python
import earthaccess

# Search needs no login. Bounding box: (west, south, east, north)
results = earthaccess.search_data(
    short_name="SRTMGL1",
    bounding_box=(-74.3, 40.5, -73.7, 40.95),  # New York City
    temporal=("2000-02-11", "2000-02-22"),
    count=5,
)
print(len(results), "granules")
for g in results:
    print(g["umm"]["GranuleUR"])

# Next step (downloads need a free Earthdata Login):
# earthaccess.login()
# files = earthaccess.download(results[:1], "./srtm")
```
