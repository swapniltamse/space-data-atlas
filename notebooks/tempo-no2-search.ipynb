# Search hourly TEMPO NO2 over a city

Dataset: [TEMPO Gridded Nitrogen Dioxide (L3) V04](https://cmr.earthdata.nasa.gov/search/concepts/C3685896708-LARC_CLOUD.html). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#tempo-no2).

```python
import earthaccess

# Search needs no login. Bounding box: (west, south, east, north)
results = earthaccess.search_data(
    short_name="TEMPO_NO2_L3",
    bounding_box=(-74.3, 40.5, -73.7, 40.95),  # New York City
    temporal=("2026-09-01", "2026-10-15"),
    count=5,
)
print(len(results), "granules")
for g in results:
    print(g["umm"]["GranuleUR"])

# Next step (downloads need a free Earthdata Login):
# earthaccess.login()
# files = earthaccess.download(results[:1], "./tempo")
# import xarray as xr; ds = xr.open_dataset(files[0])
```
