# Find VIIRS fire swaths for a region

Dataset: [VIIRS/NPP Active Fires 6-Min L2 Swath 375m (VNP14IMG) V002](https://www.earthdata.nasa.gov/data/catalog/lpcloud-vnp14img-002). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#viirs-snpp-fire-vnp14img).

```python
import earthaccess

# VIIRS 375 m active fire swaths over California, one week
results = earthaccess.search_data(
    short_name="VNP14IMG",
    bounding_box=(-124.5, 32.5, -114.0, 42.0),
    temporal=("2025-08-01", "2025-08-07"),
    count=5,
)
print(len(results), "granules")
for g in results:
    print(g["umm"]["GranuleUR"])

# Swap short_name to "MOD14" for the Terra MODIS record (2000 onward).
# Downloads need Earthdata Login:
# earthaccess.login()
# earthaccess.download(results, "./fires")
```
