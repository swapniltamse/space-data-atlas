# Search VIIRS NOAA-20 fire detections

Dataset: [VIIRS/JPSS1 (NOAA-20) Active Fires 6-Min L2 Swath 375m (VJ114IMG) V002](https://www.earthdata.nasa.gov/data/catalog/lpcloud-vj114img-002). From [Space Data Atlas](https://spacedata.swapniltamse.com/starters/#viirs-noaa20-fire-vj114img).

```python
import earthaccess

# Search needs no login. Bounding box: (west, south, east, north)
results = earthaccess.search_data(
    short_name="VJ114IMG",
    bounding_box=(-124.4, 32.5, -114.1, 42.0),  # California
    temporal=("2026-09-01", "2026-10-15"),
    count=5,
)
print(len(results), "granules")
for g in results:
    print(g["umm"]["GranuleUR"])

# Next step (downloads need a free Earthdata Login):
# earthaccess.login()
# files = earthaccess.download(results[:1], "./vj114img")
# import xarray as xr; ds = xr.open_dataset(files[0])
```
